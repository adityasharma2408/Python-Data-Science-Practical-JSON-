```python
# Step 1: Set the chunk size
chunk_size = 2000

# Step 2: Create dictionaries to store totals and counts
total_fare = {}
count_fare = {}

# Step 3: Read the CSV in chunks
for chunk in pd.read_csv("titanic dataset.csv", chunksize=chunk_size):

    # Step 4: Calculate the sum and count of fares for each class
    grouped = chunk.groupby("Pclass")["Fare"].agg(["sum", "count"])

    # Step 5: Combine the results from each chunk
    for pclass, row in grouped.iterrows():

        total_fare[pclass] = (
            total_fare.get(pclass, 0) + row["sum"]
        )

        count_fare[pclass] = (
            count_fare.get(pclass, 0) + row["count"]
        )

# Step 6: Calculate the final mean for each passenger class
for pclass in sorted(total_fare):
    mean_fare = total_fare[pclass] / count_fare[pclass]

    print("Passenger Class:", pclass)
    print("Average Fare:", round(mean_fare, 2))
```

```python
# Create a DataFrame
df = pd.DataFrame({
    "Name": ["Aman", "Rahul", "Priya"],
    "Marks": [85, 90, 78],
    "City": ["Mathura", "Delhi", "Agra"]
})

print("Original DataFrame:")
print(df)

# Export the DataFrame to JSON
df.to_json("students.json", orient="records", indent=4)

# Read the JSON file back
df2 = pd.read_json("students.json", orient="records")

print("\nDataFrame after reading JSON:")
print(df2)

# Check equality
print("\nAre both DataFrames equal?", df.equals(df2))
```

```python
import json
import pandas as pd

# Practical 1: JSON Basics

t = '''{
    "shop": "Fresh Mart",
    "open": true,
    "owner": null,
    "items": [
        {"name": "Rice", "price": 60, "qty": 10},
        {"name": "Oil", "price": 150, "qty": 4},
        {"name": "Tea", "price": 120, "qty": 6}
    ]
}'''

# Question 1: Load JSON and print type
data = json.loads(t)

print("1. Type of data:")
print(type(data))

# Question 2: Second item name and number of items
print("\n2. Second item:")
print(data["items"][1]["name"])

print("Number of items:")
print(len(data["items"]))

# Question 3: Print owner and its type
print("\n3. Owner:")
print(data["owner"])
print(type(data["owner"]))

# Question 4: Convert object back to JSON text
print("\n4. JSON text with indentation 2:")
print(json.dumps(data, indent=2))


# Practical 2: JSON to DataFrame

# Question 1: Create DataFrame
df = pd.DataFrame(data["items"])

print("\nPractical 2 - Question 1:")
print(df)

# Question 2: Add total column
df["total"] = df["price"] * df["qty"]

print("\nPractical 2 - Question 2:")
print(df)

# Question 3: Grand total and highest-total item
grand_total = df["total"].sum()

print("\nPractical 2 - Question 3:")
print("Grand Total:", grand_total)

highest_item = df.loc[df["total"].idxmax(), "name"]

print("Item with highest total:", highest_item)
```
